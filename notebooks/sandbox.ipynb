```python
import json

from bson.json_util import loads
from bondi.mongodb import MongoCollectionRef, MongoFindQuery

ejson = '{"name": {"$regex": "tom*"}}'
target = MongoCollectionRef("demo", "users")

for label, decode in [
    ("Standard JSON", json.loads),
    ("Extended JSON", loads),
]:
    predicate = decode(ejson)
    print(f"\n{label}: {predicate!r}")
    print("Value type:", type(predicate["name"]).__name__)

    query = MongoFindQuery(target, predicate)  # Both construct successfully

    try:
        serialized = query.to_json()
        print("Serialized query:", serialized)
        restored = MongoFindQuery.from_json(serialized)
        print("Round-trip succeeded:", restored.filter)
    except ValueError as error:
        print("Serialization failed:", error)
```
