```python
import matplotlib.pyplot as plt
import pandas as pd

# Create a small table and calculate a new column.
names = pd.DataFrame({"name": ["World", "Uwe", "Thomas", "Student"]})
names["name_length"] = names["name"].str.len()
display(names)

# pandas uses Matplotlib to draw the chart.
fig, ax = plt.subplots(figsize=(6, 4), dpi=150)
names.plot.bar(
    x="name",
    y="name_length",
    legend=False,
    rot=0,
    ax=ax,
)
ax.set_ylabel("Number of characters")
ax.set_title("Greeting name lengths")
plt.show()
```
